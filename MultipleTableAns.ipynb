```python
#16-Find order information given in January 2017
# Format for date is %Y-%m-%d - 2017-01-15
#strftime('%m', date(OrdDate)) gives Month-'01'
#strftime('%Y', date(OrdDate)) gives Year-'2017'
#strftime('%d', date(OrdDate)) gives Date-'15'
#For Hour, minute second, you can use  '%H:%M:%S'
txt='''
Select *
From OrderTbl
Where strftime('%m', date(OrdDate))='01' AND strftime('%Y', date(OrdDate))='2017'

'''
Q16 =  psql.sqldf(txt)
Q16
```

```python
#17-Find order and value of order given each month
# Format for date is %Y-%m-%d - 2017-01-15
#strftime('%m', date(OrdDate)) gives Month-'01'
#strftime('%Y', date(OrdDate)) gives Year-'2017'
#strftime('%Y', date(OrdDate)) gives Date-'15'
txt='''
SELECT strftime('%m', date(OrdDate)) AS Month, strftime('%Y', date(OrdDate)) AS Year
, count(Distinct OrdLine.OrdNo) as 'Number of Order', sum(Qty*ProdPrice) AS 'Total Value of Order'
FROM OrderTbl,OrdLine,Product
WHERE OrderTbl.OrdNo=OrdLine.OrdNo and OrdLine.ProdNo=Product.ProdNo
GROUP BY strftime('%m', date(OrdDate)), strftime('%Y', date(OrdDate))

'''
Q17 =  psql.sqldf(txt)
Q17
```

JOIN

image.png

Inner Join: Includes only common rows from table 1 and 2
SELECT column_name(s) FROM table1 INNER JOIN table2 ON table1.column_name = table2.column_name

Left Join : Include all rows from table 1 and matching rows from table 2
SELECT column_name(s) FROM table1 LEFT JOIN table2 ON table1.column_name = table2.column_name

Right Join : Include all rows from table 1 and matching rows from table 2
SELECT column_name(s) FROM table1 RIGHT JOIN table2 ON table1.column_name = table2.column_name

Full Outer Join : Include all rows from table 1 and table 2
SELECT column_name(s) FROM table1 FULL OUTER JOIN table2 ON table1.column_name = table2.column_name

Cross Join:A record in table1 is paired with all record on table 2
SELECT column_name(s) FROM table1,table2

NOT 1: INNER JOIN is the same as
SELECT column_name(s) FROM table1,table2 WHERE table1.column_name = table2.column_name

NOT 2: Right join and Full Outer Join is not supported in Sql lite
For right join, you can use
SELECT column_name(s) FROM table2 LEFT JOIN table1 ON table1.column_name = table2.column_name
